# Esta carpeta esta orientada a entrenar distintos modelos para comprobar la efectividad del modelo propuesto en el paper

En específico este notebook es para separar los datos

```python
import pandas as pd
```

```python
data= pd.read_csv('../datos/train_poll_v1s1.csv')
data.describe()
```

```python
#show the classes on column ai
data['ai'].value_counts()
```

Como se tratan de 460 muestras con clases balanceadas, se separaran los datos en 70% entrenamiento y 30% validacion para que todos los modelos usen la misma separación

```python
#Seleccionar los primeros 320
data_entrenamiento = data.iloc[:320, :]
data_entrenamiento['ai'].value_counts()
```

```python
#Seleccionar los ultimos 138
data_validacion = data.iloc[:138, :]
data_validacion['ai'].value_counts()
# Guardar como csv
data_validacion.to_csv('validation.csv', index=False)
```

Las subselecciones de datos no presentan clases completamente balanceadas, pero cercanas a una distribución equitativa así que se usarán para los experimentos

# Dataset pequeño (60 muestras aleatorias)

```python
# Seleccionar 30 muestras de cada clase
train_small = data_entrenamiento.groupby('ai').apply(lambda x: x.sample(30, replace=False)).reset_index(drop=True)
train_small['ai'].value_counts()
# Guardar como .csv
train_small.to_csv('train_small.csv', index=False)
```

# Dataset grande (300 muestras aleatorias)


```python
# Seleccionar 150 muestras de cada clase
train_big = data_entrenamiento.groupby('ai').apply(lambda x: x.sample(150, replace=False)).reset_index(drop=True)
train_big['ai'].value_counts()
# Guardar como .csv
train_big.to_csv('train_big.csv', index=False)
```
